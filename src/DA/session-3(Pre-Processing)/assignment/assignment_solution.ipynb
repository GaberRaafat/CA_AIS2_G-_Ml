## 3 - Inspect the dataset

`Check_data_type(df)` is a small data quality report, not only `df.dtypes`.
For every column it gives the datatype, the number of unique values, the missing values
and a small note if the column looks like a category.
The report is transposed so it is easy to read.

````python
report = Check_data_type(df, CATEGORY_LIMIT)
report
````

### Reading the report

- **Age** is `float64` and it has a lot of unique values, so it is a numeric column, and about **19.9%** of it is missing.
- **Embarked** has only **3** unique values (S, C, Q) and 2 missing rows, so it is a category.
- The columns that look like categories are: `Survived`, `Pclass`, `Sex`, `SibSp`, `Parch` and `Embarked`.
- **Cabin** is text with 147 unique values and **77%** missing, so later it will be a problem column.
