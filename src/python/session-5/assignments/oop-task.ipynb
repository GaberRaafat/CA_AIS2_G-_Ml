## Task 1 – Email Analyzer

```python
class EmailAnalyzer:
    """
    EmailAnalyzer class handles checking if an email is valid
    and extracting information from it.

    Attributes:
        email (str): The email address to be analyzed.
    """

    def __init__(self, email):
        """
        Stores the email inside the object.

        Args:
            email (str): The email address to work on.
        """
        self.email = email

    def is_valid_email(self):
        """
        Checks if the email is valid.

        Returns:
            bool: True if the email is valid, otherwise False.
        """
        # Rule 1: exactly one "@"
        if self.email.count("@") != 1:
            return False

        # Rule 2: at least one "." after the "@"
        at_position = self.email.index("@")
        after_at = self.email[at_position + 1:]

        if "." in after_at:
            return True
        else:
            return False

    def extract_username(self):
        """
        Gets the username part of the email.

        Returns:
            str: Everything before the '@'.
        """
        return self.email.split("@")[0]

    def extract_domain(self):
        """
        Gets the domain part of the email.

        Returns:
            str: The part between '@' and the last '.'.
        """
        domain_part = self.email.split("@")[1]
        return domain_part.rsplit(".", 1)[0]

    def check_domain_ending(self):
        """
        Checks what kind of domain the email has.

        Returns:
            str: 'Commercial Domain', 'Educational Domain' or 'Other Domain'.
        """
        if self.email.endswith(".com"):
            return "Commercial Domain"
        elif self.email.endswith(".edu"):
            return "Educational Domain"
        else:
            return "Other Domain"

    def analyze_email(self):
        """
        Runs the full analysis on the email.

        Returns:
            dict or str: A dictionary with the info, or 'Invalid email'.
        """
        if not self.is_valid_email():
            return "Invalid email"
        return {
            "Username": self.extract_username(),
            "Domain": self.extract_domain(),
            "Domain Type": self.check_domain_ending(),
        }


# create an object from the class
analyzer = EmailAnalyzer("gaber_raafat@gmail.com")

# call the main method using the object
result = analyzer.analyze_email()

# print the result to check the class works
print(result)

# test it also with an invalid email
analyzer2 = EmailAnalyzer("gaber_raafatgmail.com")
print(analyzer2.analyze_email())
```

## Task 2 – Message Decoder (reverse the first word + remove vowels from the second)

```python
class MessageDecoder:
    """
    MessageDecoder class handles decoding a secret message.
    It cleans the message, reverses the first word and
    removes the vowels from the second word.

    Attributes:
        message (str): The encoded message to decode.
    """

    def __init__(self, message):
        """
        Stores the encoded message inside the object.

        Args:
            message (str): The encoded message.
        """
        self.message = message

    def extract_core(self):
        """
        Keeps only letters and spaces from the message.

        Returns:
            str: The clean part of the message.
        """
        core = ""
        for ch in self.message:
            if ch.isalpha() or ch == " ":
                core = core + ch
        return core.strip()

    def reverse_word(self, word):
        """
        Reverses a word.

        Returns:
            str: The word written backwards.
        """
        return word[::-1]

    def remove_vowels(self, word):
        """
        Removes any vowel from the word.

        Returns:
            str: The word without vowels.
        """
        vowels = "aeiouAEIOU"
        result = ""
        for ch in word:
            if ch not in vowels:
                result = result + ch
        return result

    def decode_message(self):
        """
        Decodes the full message step by step.

        Returns:
            str: The decoded message.
        """
        core = self.extract_core()                  # step 1: clean the message
        words = core.split(" ")

        first_word = self.reverse_word(words[0])    # step 2: reverse first word
        second_word = self.remove_vowels(words[1])  # step 3: remove vowels

        return first_word + " " + second_word       # step 4: join them


# create an object from the class
decoder = MessageDecoder("###!!@mocleW EPGTQ!!!6789")

# call the main method using the object
decoded = decoder.decode_message()

# print the result to check the class works
print("Encoded:", decoder.message)
print("Decoded:", decoded)
```

## Task 3 – Message Decoder (reverse the first word + fix shifted vowels I -> E and O -> U)

```python
class MessageDecoderV2:
    """
    MessageDecoderV2 class handles decoding a secret message.
    It cleans the message, reverses the first word and replaces
    the shifted vowels in the second word (I -> E and O -> U).

    Attributes:
        message (str): The encoded message to decode.
    """

    def __init__(self, message):
        """
        Stores the encoded message inside the object.

        Args:
            message (str): The encoded message.
        """
        self.message = message

    def extract_core(self):
        """
        Keeps only letters and spaces from the message.

        Returns:
            str: The clean part of the message.
        """
        core = ""
        for ch in self.message:
            if ch.isalpha() or ch == " ":
                core = core + ch
        return core.strip()

    def reverse_word(self, word):
        """
        Reverses a word: 'gnirtS' -> 'String'.

        Returns:
            str: The word written backwards.
        """
        return word[::-1]

    def replace_shifted_vowels(self, word):
        """
        Replaces the shifted vowels: I -> E and O -> U.

        Returns:
            str: The word after fixing the vowels.
        """
        vowel_map = {"I": "E", "O": "U"}

        result = ""
        for ch in word:
            if ch in vowel_map:
                result = result + vowel_map[ch]   # swap it
            else:
                result = result + ch              # keep it as is
        return result

    def decode_message(self):
        """
        Decodes the full message step by step.

        Returns:
            str: The decoded message.
        """
        core = self.extract_core()                          # step 1
        words = core.split(" ")

        first_word = self.reverse_word(words[0])            # step 2
        second_word = self.replace_shifted_vowels(words[1]) # step 3

        return first_word + " " + second_word               # step 4


# create an object from the class
decoder2 = MessageDecoderV2("&&&**$gnirtS PLIO!!@1234")

# call the main method using the object
decoded2 = decoder2.decode_message()

# print the result to check the class works
print("Encoded:", decoder2.message)
print("Decoded:", decoded2)
```

## Task 5 – Multiplication Table

```python
class MultiplicationTable:
    """
    MultiplicationTable class handles printing the multiplication
    table of a number.

    Attributes:
        number (int): The number whose table we want to print.
        upto (int): How many rows to print.
    """

    def __init__(self, number, upto=10):
        """
        Stores the number and how many rows to print.

        Args:
            number (int): The number whose table we want to print.
            upto (int): How many rows to print. Defaults to 10.
        """
        self.number = number
        self.upto = upto

    def print_table(self):
        """
        Prints the multiplication table of the number.

        Returns:
            None. The table is printed to the screen.
        """
        for i in range(1, self.upto + 1):
            print(f"{self.number} * {i} = {self.number * i}")


# create an object from the class
table = MultiplicationTable(9)

# call the main method using the object (it prints the result)
table.print_table()
```

## Task 6 – Twin Primes

```python
class TwinPrimes:
    """
    TwinPrimes class finds twin prime pairs below a limit.
    Twin primes are two primes that differ by 2, such as 11 and 13.

    Attributes:
        limit (int): Print pairs where both numbers are below this value.
    """

    def __init__(self, limit=1000):
        """
        Stores the limit inside the object.

        Args:
            limit (int): The upper limit for the search. Defaults to 1000.
        """
        self.limit = limit

    def is_prime(self, number):
        """
        Checks whether a number is prime.

        Returns:
            bool: True if the number is prime, otherwise False.
        """
        if number < 2:
            return False
        for divisor in range(2, int(number ** 0.5) + 1):
            if number % divisor == 0:
                return False
        return True

    def print_twin_primes(self):
        """
        Prints all twin prime pairs below the limit.
        This method calls is_prime() inside it.

        Returns:
            None. The pairs are printed to the screen.
        """
        print("Twin Prime:")
        for number in range(2, self.limit - 2):
            if self.is_prime(number) and self.is_prime(number + 2):
                print(number, "and", number + 2)


# create an object from the class (I used 100 to keep the output short)
twins = TwinPrimes(100)

# call the main method using the object (it prints the result)
twins.print_twin_primes()
```

## Task 7 – Prime Factors

```python
class PrimeFactors:
    """
    PrimeFactors class finds the prime factors of a number.
    Example: 56 -> [2, 2, 2, 7]

    Attributes:
        number (int): A positive integer to factorise.
    """

    def __init__(self, number):
        """
        Stores the number inside the object.

        Args:
            number (int): The number to factorise.
        """
        self.number = number

    def find_factors(self):
        """
        Finds the prime factors of the number.

        Returns:
            list: The prime factors, repeated as many times as they divide.
        """
        factors = []
        number = self.number   # work on a copy so self.number stays the same
        divisor = 2
        while number > 1:
            while number % divisor == 0:
                factors.append(divisor)
                number = number // divisor
            divisor += 1
        return factors


# create an object from the class
prime_factors = PrimeFactors(56)

# call the main method using the object
factors = prime_factors.find_factors()

# print the result to check the class works
print("Prime factors of", prime_factors.number, "are:", factors)
```

## Task 8 – Decimal to Binary

```python
class DecimalToBinary:
    """
    DecimalToBinary class converts a decimal number to its
    binary representation. Example: 11 -> '1011'

    Attributes:
        number (int): A non-negative integer to convert.
    """

    def __init__(self, number):
        """
        Stores the number inside the object.

        Args:
            number (int): The number to convert.
        """
        self.number = number

    def convert(self):
        """
        Converts the number to binary.

        Returns:
            str: The binary form of the number as a string.
        """
        if self.number == 0:
            return "0"

        binary = ""
        number = self.number   # work on a copy so self.number stays the same
        while number > 0:
            binary = str(number % 2) + binary
            number = number // 2
        return binary


# create an object from the class
converter = DecimalToBinary(11)

# call the main method using the object
binary = converter.convert()

# print the result to check the class works
print(converter.number, "in binary is:", binary)
```

## Task 9 – Perfect Numbers

```python
class PerfectNumbers:
    """
    PerfectNumbers class prints every perfect number in a given range.
    A perfect number equals the sum of its proper divisors,
    for example 28 = 1 + 2 + 4 + 7 + 14.

    Attributes:
        start (int): First number of the range (included).
        end (int): Last number of the range (included).
    """

    def __init__(self, start, end):
        """
        Stores the range inside the object.

        Args:
            start (int): First number of the range.
            end (int): Last number of the range.
        """
        self.start = start
        self.end = end

    def sum_of_proper_divisors(self, number):
        """
        Adds up all proper divisors of a number.
        A proper divisor is any divisor smaller than the number itself.

        Returns:
            int: The sum of the proper divisors.
        """
        total = 0
        for i in range(1, number):
            if number % i == 0:
                total += i
        return total

    def print_perfect_numbers(self):
        """
        Prints every perfect number in the range.
        This method calls sum_of_proper_divisors() inside it.

        Returns:
            None. The perfect numbers are printed to the screen.
        """
        for number in range(self.start, self.end + 1):
            if self.sum_of_proper_divisors(number) == number:
                print(number)


# create an object from the class
perfect = PerfectNumbers(1, 100)

# call the main method using the object (it prints the result)
print("Perfect numbers between", perfect.start, "and", perfect.end, ":")
perfect.print_perfect_numbers()
```

## Bonus – Basic Calculator (from session-3)

I converted the calculator into a class too. I removed the input() loop
so the notebook can run all cells without stopping, and I test the
methods directly with numbers instead.

```python
class Calculator:
    """
    Calculator class performs the basic arithmetic operations
    on two numbers (addition, subtraction, multiplication, division).

    Attributes:
        num1 (float): The first number.
        num2 (float): The second number.
    """

    def __init__(self, num1, num2):
        """
        Stores the two numbers inside the object.

        Args:
            num1 (float): The first number.
            num2 (float): The second number.
        """
        self.num1 = num1
        self.num2 = num2

    def add(self):
        """
        Adds the two numbers.

        Returns:
            float: The result of num1 + num2.
        """
        return self.num1 + self.num2

    def subtract(self):
        """
        Subtracts the second number from the first one.

        Returns:
            float: The result of num1 - num2.
        """
        return self.num1 - self.num2

    def multiply(self):
        """
        Multiplies the two numbers.

        Returns:
            float: The result of num1 * num2.
        """
        return self.num1 * self.num2

    def divide(self):
        """
        Divides the first number by the second one.

        Returns:
            float or str: The result of num1 / num2, or an error
            message when the second number is zero.
        """
        # check the second number to avoid division by zero
        if self.num2 == 0:
            return "Error: You cannot divide by zero."
        return self.num1 / self.num2


# create an object from the class
calc = Calculator(5, 3)

# call the methods using the object and print the results
print("The result of adding", calc.num1, "and", calc.num2, "is", calc.add())
print("The result of subtracting", calc.num2, "from", calc.num1, "is", calc.subtract())
print("The result of multiplying", calc.num1, "and", calc.num2, "is", calc.multiply())
print("The result of dividing", calc.num1, "by", calc.num2, "is", calc.divide())

# test the division by zero check
calc2 = Calculator(5, 0)
print(calc2.divide())
```
