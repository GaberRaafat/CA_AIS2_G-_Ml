```python
def extract_core(message):
    """Keep only letters and spaces, drop the junk around them."""
    core = ""
    for ch in message:
        if ch.isalpha() or ch == " ":
            core = core + ch
    return core.strip()


def reverse_word(word):
    """Reverse a word: 'gnirtS' -> 'String'"""
    return word[::-1]


def replace_shifted_vowels(word):
    """Replace the shifted vowels: I -> E and O -> U."""
    vowel_map = {"I": "E", "O": "U"}

    result = ""
    for ch in word:
        if ch in vowel_map:
            result = result + vowel_map[ch]   # swap it
        else:
            result = result + ch              # keep it as is
    return result


def decode_message(message):
    core = extract_core(message)                    # step 1
    words = core.split(" ")

    first_word = reverse_word(words[0])             # step 2
    second_word = replace_shifted_vowels(words[1])  # step 3

    return first_word + " " + second_word           # step 4
```

```python
encoded = "&&&**$gnirtS PLIO!!@1234"

print("Encoded:", encoded)
print("Core:   ", extract_core(encoded))
print("Decoded:", decode_message(encoded))
```
