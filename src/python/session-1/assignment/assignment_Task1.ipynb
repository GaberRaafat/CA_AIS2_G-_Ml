```python
def is_valid_email(email):
    # Rule 1: exactly one "@"
    if email.count("@") != 1:
        return False

    # Rule 2: at least one "." after the "@"
    at_position = email.index("@")
    after_at = email[at_position + 1:]

    if "." in after_at:
        return True
    else:
        return False


def extract_username(email):
    """Everything before the '@'."""
    return email.split("@")[0]


def extract_domain(email):
    """The part between '@' and the last '.'."""
    domain_part = email.split("@")[1]
    return domain_part.rsplit(".", 1)[0]


def check_domain_ending(email):
    if email.endswith(".com"):
        return "Commercial Domain"
    elif email.endswith(".edu"):
        return "Educational Domain"
    else:
        return "Other Domain"


def analyze_email(email):
    if not is_valid_email(email):
        return "Invalid email"
    return {
        "Username": extract_username(email),
        "Domain": extract_domain(email),
        "Domain Type": check_domain_ending(email),
    }


```

```python
email = "gaber_raafat@gmail.edu"

result = analyze_email(email)
print(result)
```

```python
email = "gaber_raafat@gmail.com"

result = analyze_email(email)
print(result)
```
