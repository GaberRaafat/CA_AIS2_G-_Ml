ou are given two non-empty linked lists representing two non-negative integers. The digits are stored in reverse order, and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list.

You may assume the two numbers do not contain any leading zero, except the number 0 itself.

```python
from typing import Optional


class ListNode:
    """One box in a linked list: a digit plus a pointer to the next box."""

    def __init__(self, val: int = 0, next: Optional["ListNode"] = None) -> None:
        self.val = val
        self.next = next
```

```python
class Solution:
    def addTwoNumbers(
        self, l1: Optional[ListNode], l2: Optional[ListNode]
    ) -> Optional[ListNode]:
        """Add two numbers stored as reversed linked lists.

        Args:
            l1: Head of the first linked list, digits in reverse order.
            l2: Head of the second linked list, digits in reverse order.

        Returns:
            Head of a new linked list holding the sum, also in reverse order.
        """
        # A placeholder box so we always have something to attach to.
        dummy = ListNode()
        current = dummy
        carry = 0

        # Keep going while either list has digits left, or a carry is pending.
        while l1 is not None or l2 is not None or carry > 0:

            # Take the digit if the list still has one, otherwise use 0.
            digit1 = 0
            if l1 is not None:
                digit1 = l1.val

            digit2 = 0
            if l2 is not None:
                digit2 = l2.val

            total = digit1 + digit2 + carry
            carry = total // 10        # what we pass to the next column
            digit = total % 10         # what stays in this column

            # Attach a new box holding this digit, then step forward.
            current.next = ListNode(digit)
            current = current.next

            if l1 is not None:
                l1 = l1.next
            if l2 is not None:
                l2 = l2.next

        # dummy was just a starting handle; the real answer begins after it.
        return dummy.next
```

```python
def buildList(digits: list[int]) -> Optional[ListNode]:
    """Turn a Python list of digits into a linked list.

    Args:
        digits: The digits in the order they should appear in the chain.

    Returns:
        Head of the new linked list, or None if the input is empty.
    """
    dummy = ListNode()
    current = dummy
    for digit in digits:
        current.next = ListNode(digit)
        current = current.next
    return dummy.next


def toPythonList(head: Optional[ListNode]) -> list[int]:
    """Turn a linked list back into a Python list so we can print it.

    Args:
        head: Head of the linked list.

    Returns:
        A list of the values in order.
    """
    values = []
    while head is not None:
        values.append(head.val)
        head = head.next
    return values


l1 = buildList([2, 4, 3])   # 342
l2 = buildList([5, 6, 4])   # 465
result = Solution().addTwoNumbers(l1, l2)
print(toPythonList(result))  # [7, 0, 8]  ->  807
```
