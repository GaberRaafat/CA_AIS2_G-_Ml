```python
def multiplicationTable(number: int, upto: int = 10) -> None:
    """Print the multiplication table of a number.

    Args:
        number: The number whose table we want to print.
        upto: How many rows to print. Defaults to 10.

    Returns:
        None. The table is printed to the screen.
    """
    for i in range(1, upto + 1):
        print(f"{number} * {i} = {number * i}")


multiplicationTable(9)
```

```python
def isPrime(number: int) -> bool:
    """Check whether a number is prime.

    Args:
        number: The number to test.

    Returns:
        True if the number is prime, otherwise False.
    """
    if number < 2:
        return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True


def twinPrimes(limit: int = 1000) -> None:
    """Print all twin prime pairs below a limit.

    Twin primes are two primes that differ by 2, such as 11 and 13.
    This function calls isPrime() inside it.

    Args:
        limit: Print pairs where both numbers are below this value.

    Returns:
        None. The pairs are printed to the screen.
    """
    print("Twin Prime:")
    for number in range(2, limit - 2):
        if isPrime(number) and isPrime(number + 2):
            print(number, "and", number + 2)


twinPrimes(1000)
```

```python
def primeFactors(number: int) -> list[int]:
    """Find the prime factors of a number.

    Example:
        primeFactors(56) -> [2, 2, 2, 7]

    Args:
        number: A positive integer to factorise.

    Returns:
        A list of prime factors, repeated as many times as they divide.
    """
    factors: list[int] = []
    divisor = 2
    while number > 1:
        while number % divisor == 0:
            factors.append(divisor)
            number = number // divisor
        divisor += 1
    return factors


primeFactors(56)
```

```python
def decToBin(number: int) -> str:
    """Convert a decimal number to its binary representation.

    Example:
        decToBin(11) -> "1011"

    Args:
        number: A non-negative integer.

    Returns:
        The binary form of the number as a string.
    """
    if number == 0:
        return "0"
    binary = ""
    while number > 0:
        binary = str(number % 2) + binary
        number = number // 2
    return binary


print(decToBin(11))
```

```python
def sumOfProperDivisors(number: int) -> int:
    """Add up all proper divisors of a number.

    A proper divisor is any divisor smaller than the number itself.

    Args:
        number: The number to inspect.

    Returns:
        The sum of the proper divisors.
    """
    total = 0
    for i in range(1, number):
        if number % i == 0:
            total += i
    return total


def perfectNums(start: int, end: int) -> None:
    """Print every perfect number in a given range.

    A perfect number equals the sum of its proper divisors,
    for example 28 = 1 + 2 + 4 + 7 + 14.
    This function calls sumOfProperDivisors() inside it.

    Args:
        start: First number of the range (included).
        end: Last number of the range (included).

    Returns:
        None. The perfect numbers are printed to the screen.
    """
    for number in range(start, end + 1):
        if sumOfProperDivisors(number) == number:
            print(number)


perfectNums(0, 100)
```
